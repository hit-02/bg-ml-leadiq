```
from snowflake.snowpark.context import get_active_session
import pandas as pd, numpy as np
session = get_active_session()

features = ['VIEWED_THIS_BOAT','CLICKED_GALLERY','SAVED_BOAT','HAS_PHONE','IS_NEW_VISITOR',
            'LISTING_PRICE','USER_LISTING_DISTANCE','HAS_TIER','SAME_STATE','CROSS_COUNTRY',
            'MENTIONS_SURVEY','MENTIONS_SEA_TRIAL',
            'SRC_BT_PORTAL','SRC_BT_APP','SRC_YW_WEB','SRC_YW_APP','SRC_BOATS_COM',
            'SRC_SOCIAL','SRC_MLS','SRC_OTHER',
            'N_PRIOR_LEADS','N_DISTINCT_LISTINGS_PRIOR','DAYS_SINCE_LAST_LEAD','PRIOR_BOAT_OWNER']

df = session.sql(f"SELECT {', '.join(features)} "
                 f"FROM AWS_DP_PROD.ML_OPS.LEADIQ_TRAINING_SET_V4").to_pandas()


df['LISTING_PRICE'] = df['LISTING_PRICE'].clip(0, df['LISTING_PRICE'].quantile(0.995))

corr = df.corr(method='spearman')   
```

```
import matplotlib.pyplot as plt, seaborn as sns

fig = plt.figure(figsize=(14, 12))
sns.heatmap(corr, cmap='coolwarm', center=0, vmin=-1, vmax=1,
            annot=True, fmt='.2f', annot_kws={'size': 5},
            cbar_kws={'label': 'Spearman correlation'})
plt.title('Lead IQ features correlation heatmap')
plt.tight_layout()
plt.show()         
plt.savefig('feature_heatmap.png', dpi=150)   
```
